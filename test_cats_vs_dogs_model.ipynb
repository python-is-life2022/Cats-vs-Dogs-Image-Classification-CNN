```python
import os
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow import keras
from keras.models import load_model
```

```python
model = load_model('cat_vs_dog_images_classification_CNN_model (2).keras')
```

```python
test_images = os.listdir('test_photos/test_photos')
for img in test_images:
    image = plt.imread(f'test_photos/test_photos/{img}')
    image = tf.image.resize(image, (64, 64))
    image = tf.cast(image, tf.float32) / 255.0
    image = tf.expand_dims(image, axis= 0)
    y_hat = model.predict(image)
    if y_hat > 0.5:
        print('Dog')
    else:
        print('Cat')
```

```python
image = plt.imread(f'test_photos/test_photos/1.jpg')
plt.imshow(image)
image = tf.image.resize(image, (64, 64))
image = tf.cast(image, tf.float32) / 255.0
image = tf.expand_dims(image, axis= 0)
y_hat = model.predict(image)
if y_hat > 0.5:
    plt.title('Dog')
else:
    plt.title('Cat')
```
